```python
from src.data_loader import load_data
```

```python
sample_df = load_data(
    "../data/raw/reviews_final_for_db.csv"
)

sample_df.head()
```

```python
bank_mapping = {

    1: "Commercial Bank of Ethiopia",

    2: "Bank of Abyssinia",

    3: "Dashen Bank"
}

sample_df["bank"] = (
    sample_df["bank_id"]
    .map(bank_mapping)
)
```

```python
sample_df.columns
```

```python
sentiment_distribution = sample_df.groupby(
    ["bank", "sentiment_label"]
).size().reset_index(name="count")

sentiment_distribution
```

```python
pivot_df = sentiment_distribution.pivot(
    index="bank",
    columns="sentiment_label",
    values="count"
)

pivot_df
```

```python
import matplotlib.pyplot as plt

pivot_df.plot(kind="bar")

plt.title("Sentiment Distribution by Bank")

plt.xlabel("Bank")

plt.ylabel("Number of Reviews")

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

```python
average_rating = sample_df.groupby(
    "bank"
)["rating"].mean().reset_index()

average_rating
```

```python
average_rating.plot(
    x="bank",
    y="rating",
    kind="bar"
)

plt.title("Average Rating by Bank")

plt.xlabel("Bank")

plt.ylabel("Average Rating")

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

```python
theme_counts = sample_df.groupby(
    ["bank", "identified_theme"]
).size().reset_index(name="count")

theme_counts.sort_values(
    by="count",
    ascending=False
).head(15)
```

```python
top_themes = theme_counts.sort_values(
    by="count",
    ascending=False
).head(10)

top_themes.plot(
    x="identified_theme",
    y="count",
    kind="bar"
)

plt.title("Top Themes Across Banks")

plt.xlabel("Theme")

plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
